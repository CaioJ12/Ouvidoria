# Entrega 1 — Comparação de Representações

**Proposta individual: Ouvidoria Semântica**

Este notebook compara BoW, TF-IDF e dois modelos de embeddings multilíngues nos três pares definidos pelo desafio. Os textos e categorias são sintéticos e foram construídos especificamente para a atividade.

```python
from pathlib import Path
import sys
import pandas as pd
import numpy as np
sys.path.insert(0, str(Path.cwd()))
from ouvidoria import carregar, comparar_pares, MODELOS

df = carregar()
print(df.shape)
df.head()
```

```python
tabelas = []
embeddings = {}
for nome in MODELOS:
    tabela, emb = comparar_pares(df, nome)
    tabela["Modelo"] = nome
    tabelas.append(tabela)
    embeddings[nome] = emb

comparacao = pd.concat(tabelas, ignore_index=True)
display(comparacao.round(4))
Path("resultados").mkdir(exist_ok=True)
comparacao.to_csv("resultados/comparacao_representacoes.csv", index=False)
```

```python
import matplotlib.pyplot as plt
tabela = comparacao.copy()
fig, ax = plt.subplots(figsize=(10,5))
pivot = tabela.pivot(index="Par", columns="Modelo", values="Embeddings")
comparacao.groupby("Par")[["BoW","TF-IDF"]].mean().plot.bar(ax=ax)
ax.set_ylim(0,1)
ax.set_ylabel("Cosseno")
ax.set_title("Comparação das representações esparsas")
fig.tight_layout()
fig.savefig("resultados/comparacao_esparsas.png", dpi=160)
plt.show()
```

## Como interpretar

BoW depende de coincidências literais e, por isso, pode atribuir pouca similaridade a paráfrases. TF-IDF reduz o peso de termos muito comuns, mas continua limitado ao vocabulário observado. Embeddings densos procuram representar significado e podem aproximar frases com palavras diferentes. Ainda assim, similaridade semântica não significa identidade de ocorrência.

Para o par M008 × M031, espera-se menor similaridade do que nos dois pares tematicamente relacionados, pois o primeiro trata de atendimento médico e o segundo de iluminação em praça. A conclusão quantitativa deve ser feita a partir da tabela produzida na execução.
